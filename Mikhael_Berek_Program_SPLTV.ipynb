```python
# PROGRAM PENYELESAIAN SPLTV
# Menggunakan metode matriks dengan NumPy

# Cell 1: Mengimpor library NumPy
import numpy as np

# Cell 2: Membuat matriks koefisien
A = np.array([
    [2, 1, 1],
    [1, 2, 1],
    [1, 1, 2]
])

# Cell 3: Membuat matriks hasil
B = np.array([10, 11, 12])

# Cell 4: Menghitung nilai x, y, dan z
hasil = np.linalg.solve(A, B)

# Cell 5: Menampilkan hasil
print("Hasil Penyelesaian SPLTV:")
print("Nilai x =", hasil[0])
print("Nilai y =", hasil[1])
print("Nilai z =", hasil[2])
```

```python

# CELL 2: MEMERIKSA KEBENARAN HASIL

# Menghitung kembali hasil dari ketiga persamaan
hasil_cek = np.dot(A, hasil)

# Menampilkan hasil pemeriksaan
print("Pemeriksaan Hasil SPLTV:")

print("Persamaan 1 =", hasil_cek[0])
print("Persamaan 2 =", hasil_cek[1])
print("Persamaan 3 =", hasil_cek[2])

# Memeriksa apakah hasil sesuai dengan matriks B
if np.allclose(hasil_cek, B):
    print("Kesimpulan: Hasil perhitungan BENAR.")
else:
    print("Kesimpulan: Hasil perhitungan SALAH.")
```

```python

# CELL 3: MENGHITUNG DETERMINAN MATRIKS

# Menghitung nilai determinan matriks A
determinan = np.linalg.det(A)

# Menampilkan nilai determinan
print("Nilai Determinan Matriks:", determinan)

# Memeriksa jenis penyelesaian SPLTV
if abs(determinan) > 1e-10:
    print("Sistem memiliki satu penyelesaian tunggal.")
else:
    print("Sistem tidak memiliki penyelesaian tunggal.")
```

```python

# CELL 4: MENAMPILKAN HASIL DALAM BENTUK TABEL

# Menampilkan judul tabel
print("TABEL HASIL PENYELESAIAN SPLTV")
print("==============================")

# Menampilkan nilai setiap variabel
print("{:<15} {:<15}".format("Variabel", "Nilai"))

print("{:<15} {:<15.2f}".format("X", hasil[0]))
print("{:<15} {:<15.2f}".format("Y", hasil[1]))
print("{:<15} {:<15.2f}".format("Z", hasil[2]))

print("==============================")
print("Perhitungan selesai.")
```

```python

# CELL 5: KESIMPULAN AKHIR PROGRAM SPLTV

print("====================================")
print("       KESIMPULAN PROGRAM SPLTV")
print("====================================")

print("Sistem Persamaan Linear Tiga Variabel")
print("memiliki hasil penyelesaian sebagai berikut:")

print("Nilai X =", round(hasil[0], 2))
print("Nilai Y =", round(hasil[1], 2))
print("Nilai Z =", round(hasil[2], 2))

print("------------------------------------")
print("Determinan Matriks =", round(determinan, 2))

if abs(determinan) > 1e-10 and np.allclose(np.dot(A, hasil), B):
    print("Kesimpulan: Sistem memiliki satu solusi.")
    print("Program berhasil menyelesaikan SPLTV.")
else:
    print("Kesimpulan: Sistem tidak memiliki solusi tunggal.")

print("====================================")
```
